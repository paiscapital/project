# Public Relatons For NGO

PR is literally the bridge between the NGO and society.

## 1: Most Valuable Asset For NGO
For companies, most valuable asset:

    Money + Products + Brand

For NGOs, most valuable asset:

    TRUST

Trust is their currency:

    People donate because they trust. 
    Volunteers join because they trust.
    Government cooperates because they trust.
    Media covers them because they trust them.

No trust = No NGO. That is why NGO PR spends huge effort protecting reputation.

## 2: Advocacy
Advocacy means influencing people or governments to support a cause. Instead of saying: 

    "Buy our product."

NGO says

    "Please stop child labour."
            or

    "Protect forests."
            or
    "Support mental health."
The goal is changing attitudes, behaviour, or laws, not selling products. Today NGOs use:

    Instagram
    TikTok
    Facebook
    YouTube
    LinkedIn

Social media helps them: 
1. Raise awareness.
2. Recruit volunteers.
3. Collect donations.
4. Fight misinformation.
5. Mobilize supporters.

One viral post can reach millions.  

## 3: Characteristics
#### a) Reliable Message
People must believe the NGO. Example:

    "We built 10 schools."
            Not
    "We are trying our best."
Evidence builds credibility.

#### b) Social Adhesion
Getting society to support the objective. Example,  
Before:

    Nobody cares about recycling.
After campaign:

    Everyone starts recycling.
Society "sticks" to the idea.

#### c) Policy Influence
Many NGOs don't just help people. They change laws. Example:

    Road safety NGO
Instead of only helping accident victims,

    they lobby government
            ↓
    Government makes seatbelt law.
This creates much larger impact.

#### d) Crisis comes from outside
NGO crisis is usually external, Example:

    War.
    Political conflict.
    Government disagreement.
    Natural disasters.

Their mission itself creates controversy.

#### e) Encourage Participation
The public should not only donate. They should also:

    Volunteer.
    Share.
    Join campaigns.
    Speak up.

Become long-term supporters.

## 4: Function
#### a) Reputation Management
PR constantly asks:

    Do people trust us?
    Do donors trust us?
    Do media trust us?
    Do government trust us?
Every campaign either builds or destroys reputation.

#### b) Advocacy (pro to objective)
NGO PR influences policy, Professional activities include:

    Writing policy papers
    Meeting ministers
    Giving expert advice
    Speaking at hearings
    Lobbying ethically
This is called Public Affairs. Corporate PR rarely does this.

#### c) Fundraising (human remeber stories)
No money = No mission. PR supports fundraising by telling powerful stories. Notice:

    People donate to stories,
    not statistics.

Instead of saying:

    "10,000 children need food."
They tell
    
    "Ali, age 7, finally ate his first proper meal after three days."
Humans remember stories.

#### d) Crisis Communication
Imagine an earthquake. NGO PR immediately communicates:

    Where to get help.
    Where donations go.
    Safety information.
    Rumours.
    Volunteer instructions.
    Media updates.
Good crisis communication reduces panic.

#### c) Internal Security
PR isn't only talking to media. They also collect incident reports. Example:

    Aid workers attacked.
    Kidnapping.
    Disease outbreak.
    Political violence.
PR works with security teams to improve future safety.

## 5: Stakeholders
Anyone affected by the NGO. Examples:

    Donors
    Volunteers
    Government
    Communities
    Media
    Corporate sponsors
    Beneficiaries
Every stakeholder wants different information.

## 6: Different Donors Need Different PR
#### a) Individual Donor
Example:

    You donate RM20.
    You want emotional stories.
    You want to know
    "I helped someone."

#### b) Corporate donor
A company gives RM1 million. They also want recognition. Example:

    "This classroom is sponsored by Company X."

#### c) Foundation
They care about reports. They ask:

    Where did every ringgit go?
They demand accountability.

#### d) Legacy donor
Someone leaves property after death. Communication must be respectful and personal.

#### e) Volunteers
Money isn't the only resource, Volunteers are also assets. A Good PR:

    Recruits them.
    Matches skills.
    Motivates them.
    Retains them.
Recognition is important because people volunteer for purpose, not salary.

#### f) Strategic Partnerships
Bad partnership:

    Company gives money.
    Done.
Good partnership

    Company and NGO solve problems together.
Example:

    Solar company
    Climate NGO
        ↓
    Install renewable energy in villages.
Both achieve goals, This is a strategic partnership.

## 7: Storytelling
This is probably the strongest NGO communicationn tool, Because emotion moves people. Brain science says:

    Humans remember stories much longer than numbers.

Example, Bad:

    "500 families affected."

Good

    "Fatimah carried her baby through floodwater for six hours."

Stories create empathy.

## 8: Lobbying
Lobbying does not automatically mean corruption, Professional lobbying means:

    Meeting lawmakers.
    Presenting research.
    Suggesting better policies.
    Giving expert opinions.
Example:

    Health NGO
        ↓
    Convince government to ban smoking in restaurants.
That is advocacy through lobbying.

## 9: Cultural Sensitivity
Imagine an NGO from Europe entering Malaysia. If they ignore local culture, their campaign fails. Example:

    Advertising that is acceptable in Europe
            ↓
    May offend Malaysian culture.

Good NGO PR studies:

    Religion.
    Language.
    Traditions.
    Values.
    Local customs.
This increases campaign effectiveness.

## 10: NGO Theories
#### a) Systems Theory
NGO is connected to everything:

    Government.
    Media.
    Public.
    Donors.
    Community.

If one relationship changes, the whole system changes.

#### b) Stakeholder Theory
Not every stakeholder has equal importance. PR identifies: 

    Who has highest influence?
    Who needs most attention?
Then customizes communication.

#### c) Diffusion of Innovation Theory
Explains how new ideas spread. Example:

    First 10 people recycle.
        ↓
        100
        ↓
        1,000
        ↓
    Whole city.
NGOs use this theory for behaviour change campaigns.

#### d) Cultural Intelligence (CQ)
CQ means Being smart across cultures. Four parts:

    Knowledge
    Awareness
    Motivation
    Behaviour
Professional NGO communicators adjust communication based on local culture.
