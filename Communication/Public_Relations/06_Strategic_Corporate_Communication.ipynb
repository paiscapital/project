# Strategic Corporate Communication

Corporate Communication is the department that makes sure everyone in the company communicates with the same message so people trust and respect the company.  
Customer judge company as a one entity.  

#### The hidden formula

    Actions (What company do)
        ↓
    Communication (What the company says)
        ↓
    Perception (What people think)
        ↓
    Trust (Belief over time)
        ↓
    Reputation (Long-term judgement)
        ↓
    Business Success (Outcome)
Corporate Communication manages this entire chain.

#### Reputation is everything
Good product is not successful company. Example:

    Company A
        Best product
        Bad reputation
            ↓
        People hesitate to buy.

    Company B
        Average product
        Excellent reputation
            ↓
        People trust it.

Which sells more? Often Company B. This is why reputation becomes an economic asset. Companies spend billions protecting it.

## 1: Stakeholder
Who are stakeholders? A stakeholder is anyone who can affect the company or anyone affected by the company. Example: 

    Employees: Need salary.
    Customers: Need products.
    Investors: Need profits.
    Government: Needs taxes.
    Suppliers: Need payment.
    Communities: Need responsible behaviour.
    Media: Need information.
    NGOs: Need accountability.
Every one of these groups wants different things. Corporate Communication balances them.

## 2: Three level of communication
Level 1: Operational.  
Example
    
    Replying to customer emails.

Level 2: Tactical.  
Example

    Planning a product launch campaign.

Level 3: Strategic.  
Example

    Designing how every communication supports long-term trust, reputation, and organizational goals.

Corporate Communication operates mainly at Level 3.

## 3: The Integration Principle
a) Management Communication
    
    Leadership speeches
    Vision
    Strategy
    Decision announcements

b) Marketing Communication

    Advertising
    Promotions
    Product campaigns
    Customer communication
    
c) Company Communication
    
    Policies
    Annual reports
    Internal announcements
    Operational updates

Without integration, each department could send conflicting messages. With integration, every message reinforces the same organizational identity.

## 4: View
CC is one that have role to manage all company communication to protect trust, imej, reputation etc. The structure:

1. The Corporate Communication Manager (Decide)
2. Public relations
3. Marketing Communication
4. Internal Communication
5. Investor Relations
6. CSR
7. Crisis Communication

#### The role

    PR = external signals (radio to outside world)
    Marketing = attracting customers (sales signal)
    Internal comms = crew coordination
    IR = financial engine room
    CSR = moral compass
    Crisis team = emergency response syste
