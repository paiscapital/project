# Sports Public Relations

Sports PR = Managing people's emotions and perceptions about sports (Shared identity).    
Sports involve huge emotions, Fans don't simply buy products. They:

    cry
    celebrate
    become angry
    defend athletes
    fight online
    remember moments forever
Therefore Sports PR is much harder, A single mistake becomes worldwide news within minutes. Sports PR manages:

    athlete image
    fan emotions
    media
    sponsors
    scandals
    victories
    defeats
all at once. Sports PR is actually a Reputation Risk Management System. It tries to maximize Positive Reputation while minimizing Reputation Risk.


    Positive stories
        ↓
    Trust
        ↓
    Loyal fans
        ↓
    Sponsors
        ↓
    Revenue

    - Whereas

    Scandal
        ↓
    Loss of trust
        ↓
    Sponsors leave
        ↓
    Fans become angry
        ↓
    Revenue decreases

## 1: Scope
#### a) Internal Communication

Inside organization, Communicating with:

    athletes
    coaches
    management
    staff
Everyone must communicate consistently.

#### b External Communication
Communicating with:

    fans
    journalists
    TV
    newspapers
    social media
    government
PR ensures everyone receives the same message.

#### c) Crisis Management
This is probably the hardest job. Examples:

    doping
    corruption
    racism
    match fixing
    athlete arrested
    coach fired
PR acts immediately before rumors spread.

#### d) Promotion
Helping events become famous. Examples:

    Olympics
    World Cup
    SEA Games
    Football league

PR increases:
 
    attendance
    excitement
    publicity


#### d) Community Relations
Helping society. Examples:

    Free football clinics
    Charity matches
    School visits
    Environmental campaigns

Purpose: People don't just support your team, They love your organization.

## 2: Reputation is Everything
Imagine two athletes:

    Athlete A
    World champion
    but constantly involved in scandals.

    Athlete B
    Slightly weaker
    but respected
    kind, humble, professional.
Sponsors often prefer Athlete B, Because sponsors buy reputation, Not just performance.    
Reputation = Performance × Character × Communication    
If communication fails, people misunderstand. If character fails, trust disappears. If performance fails, incerest decreases.
Sports PR protects all three.

## 3: Fan Loyalty
Normal customer Buys phone, Next year they Changes brand. Sports fan Supports club for 10 years/20 years, Sometimes entire lifetime. Even after losing championships. This emotional attachment is called fan loyalty.   
Behavioral science explains this. Fans develop:

    identity ("I am a Liverpool fan")
    belonging
    memories
    traditions
    friendships
    emotions
The club becomes part of who they are. That's why sports PR focuses on relationships rather than one-time sales.  

## 4: SPR Cycle
    Win game
        ↓
    Positive media
        ↓
    Fans happy
        ↓
    Sponsors interested
        ↓
    Revenue increases
        ↓
    Better players
        ↓
    More wins
        ↓
    More positive PR

Negative cycle:

    Lose
    ↓
    Scandal
        ↓
    Negative media
    ↓
    Fan anger
        ↓
    Sponsors leave
    ↓
    Less money
        ↓
    Poor performance
    ↓
    More criticism
Sports PR tries to break the negative cycle early.
