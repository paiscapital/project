## Credibility-Control Paradox
The more control an organization has over its message, the less credible the message is often perceived to be. Conversely, the less control the organization has over the message, the more credible it tends to be.

Example you say : 
    
    "My restaurant is the best."
    People think:
        "Of course you'll say that."
    
    Now imagine a respected food critic says:
        "This restaurant is excellent."

    Same message. Different credibility.

    Advertising: High control, Low credibility
    PR: Low control, High credibility

## Earned Media Is Network Effects
Earned media is when other people talk about your company for free. Example, Company posts on Facebook:
    
    100 people see it.
    Now imagine: BBC reports it. Millions see it.
    
    Then:
        YouTubers discuss it.
        TikTok reacts.
        Reddit debates.
        News quotes TikTok.
        Investors discuss it.
    
    Now one story spreads across many networks.
PR professionals engineer these information.

## Reputation Management Is Bayesian Updating
Bayesian updating means changing your belief when you receive new evidence.  
Reputation Management = Peoples Opinion Keeps Updating. Example:

    Day 1:
        A company donates RM1 million to charity.
        People think: "Maybe this company is good."
        Trust = 60%

    Day 30:
        The company is caught polluting a river.
        People think: "Maybe they aren't as good as we thought."
        Trust = 35%

    Day 60:
        The CEO apologizes, cleans the river, and changes company policies.
        People think: "They're taking responsibility."
        Trust = 55%

## CSR (Corporate Social Responsiblity) Is Not Charity
CSR is the strategic management of a company's impact on society while strengthening long-term business performance. How can a company make money without harming and ideally improving society.  

Example, Company reduces pollution:

    Environmental groups support them.
    Investors like ESG.
    Government approves permits faster.
    Employees become proud.
    Customers stay loyal.
    
The company benefits while society benefits. Good CSR aligns business success with stakeholder value.

## Greenwashing
Making people believe your company is environmentally friendly when it really isn't (Lying with a green image).  

    Suppose company says : "We love nature."
    Reality: Factory pollutes rivers.
    Now journalists investigate.
    Result:

        Reputation collapses.
        Consumers become angry.
        Investors lose confidence.

This is why modern CSR requires measurable evidence, not slogans.

## Crisis Communication Is Decision Science
Decision science is the study of how to choose the best action when you don't know everything: 

    Imagine you're playing chess.
    Your opponent makes a move.
    You don't know what they'll do next.
    You don't have complete information.
    Yet you must decide your next move.
That is decision science.    

Most people think Crisis happens, Write apology, Done. Professionals think differently. A crisis is an optimization problem, Variables include:

    Time
    Trust
    Money
    Media
    Politics
    Legal risk
    Social media
    Stakeholders
    Information quality
    
Every hour changes the outcome, The crisis team asks:

    Who speaks?
    What evidence exists?
    What is uncertain?
    Should CEO appear?
    Should we apologize?
    Should we compensate?
    Should we recall products?
This is decision-making under uncertainty.

## POEM Framework
Paid, Owned, Earned Media (POEM).
This framework = how a company distributes information:

    Paid Media: You pay for visibility
        Ads (Google, TikTok, TV)
        Sponsored posts
        Influencer sponsorships
        Display banners

        Goal: instant attention + traffic

    Owned Media: You control the platform
        Company website
        Blog
        Email newsletters
        Official social media pages
        Mobile apps

        Goal: build direct relationship + long-term control

    Earned Media:
        Others talk about you
        News articles
        Viral posts
        Reviews
        Word of mouth
        Reddit / TikTok discussions

        Goal: trust + credibility
Paid Media → Owned Media → Earned Media → increases trust → improves Paid Media performance → reduces cost.

## Advanced case
#### 1: Patagonia
Most companies advertise: "Buy our jacket."   
Patagonia said: "Don't buy this jacket."  

Why? Because their real objective wasn't selling jackets, It was proving:

    "We genuinely care about sustainability."
That authenticity generated massive media coverage, stronger trust, and eventually higher sales.

#### 2: Johnson & Johnson
During the 1982 Tylenol poisoning crisis:

    Instead of protecting profits.
    They recalled products immediately.

    Short-term:
            Huge financial loss.
    Long-term:
        Public trust increased.

    Lesson:
        Protect trust before profit.
Profit returns because trust survives.
