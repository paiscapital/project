# Crisis Communication

Crisis = A sudden event that threatens an organization’s reputation, operations, or survival, and requires immediate response.    
Crisis Comm = The way an organization talks to the public during a crisis to reduce damage and maintain trust.  Its not just giving information, its:

    controlling narrative
    reducing panic
    showing responsibility
    protecting reputation
#### The different between Crisis management
| Crisis Management                      | Crisis Communication     |
| -------------------------------------- | ------------------------ |
| Fixes the problem physically           | Fixes perception & trust |
| Operational response                   | Public messaging         |
| Example: repair system, rescue victims | Example: press statement |

#### b) Preparation (before Crisis Hits but Risk exist)
Goal: be ready.  
What companies do:

    crisis plan (manual)
    spokesperson training
    media simulation
    crisis team setup
Think: "If crisis happens, we are ready in 5 minutes"

#### c) Response (During crisis)
Goal: control damage now.  
What companies must do:

    respond fast (first 1–2 hours critical)
    give clear statement
    show empathy
    take responsibility if needed

This is where reputation is saved or destroyed.

Goal: rebuild trust.  
Actions:

    compensation
    improvements
    public apology campaigns
    transparency reports

Think: "We are fixing the system"

## 3: Situational Crisis Communication Theory (SCCT)
"The way you respond depends on how much blame the public assigns to you" - Timothy Coombs    
More blame = more apology + more action.  

### 3 Crisis Types in SCCT
#### a) Victim Crisis (Low blame)
Company is also a victim

    natural disaster
    terrorism
    hacked system
Strategy:

    sympathy (Show compassion and acknowledge peoples suffering)
    support victims (Tell people what the organization is actually doing to help)
    minimal apology needed

Example, earthquake (Your hotel is included):

    Symphaty: "Our thoughts are with everyone affected by today's earthquake."
    Support Victims: "Our first priority is ensuring the safety of our guests and employees. Emergency teams are on site, and we are providing temporary accommodation and assistance to everyone affected"
    Minimal Apology: "We are sorry that our guests have experienced this difficult situation"

Easy Formula:

    Sympathy: "Our thoughts are with everyone affected..."
    Priority: "Our first priority is the safety of..."
    Action: "We have activated our emergency response team..."
    Support: "We are providing accommodation, food, transport, and assistance..."
    Updates: "We will continue to provide verified updates as more information becomes available."

#### b) Accidental Crisis (medium blame)
The organization caused the problem, but there was no intention to cause harm.

    system failure 
    technical error
    unintentional mistake

Strategy:

    apology (Recognition, Emphaty, apology, responsiblity)
    explanation
    corrective action

Example, food dilivery system failure:

    Apology: "We sincerely apologize to all customers who were affected by the duplicate charges. We understand how frustrating and concerning this situation is, and we are truly sorry for the inconvenience"
    Explaination: "The duplicate charges were caused by an unexpected software error that occurred during a scheduled payment system update earlier today. Once the issue was identified, we immediately stopped the affected payment process"
    Corrective Action: "All duplicate charges will be automatically refunded within 24 hours. We have also suspended the affected software update, added additional payment verification checks, and are conducting a full review of our payment system to prevent this from happening again"

#### c) Preventable Crisis (high blame)

    negligence
    cover-up
    unethical behavior

Strategy:

    full apology (Acknowledges the harm, expresses empathy, and commits to fixing the problem.)
    accept responsibility
    compensation (Apologies don't pay hospital bills.)
    leadership accountability

Example, airline:

    Full Apology: "We are deeply sorry that our products caused illness to our customers. The safety of our customers should always come first, and in this case we failed to meet that responsibility. Our thoughts are with everyone affected, and we are working closely with health authorities to support them and ensure this never happens again"
    Accept responsiblity: "Although several factors contributed to this incident, the responsibility ultimately belongs to FreshBite. We failed to stop production after receiving safety warnings, and that decision was wrong"
    Conpensation: "We will pay every medical bill and compensate everyone affected"
    Leadership accountablity: "As CEO, I accept full responsibility for this failure. The decision to continue production should never have happened under my leadership"

## 4: Reputation Protection
Reputation is destroyed when:

    silence happens too long
    denial without proof
    emotion is ignored
    inconsistency in messaging

#### a) Speed (Fast response)
First statement in hours, not days

#### b) Transparency
Dont hide facts

#### c) Empathy
Show human concern first, not corporate logic.  
Example:
    
    ❌ "We are investigating"
    ✔ "We are deeply sorry and our priority is safety"

#### d) Consistency
Same message across:

    CEO
    PR team
    social media
    press

#### c) Accountability
Even if unsure, show responsibility:

    "We are taking full responsibility for investigation"

## 5: Media Response Strategy
Media is not your enemy. It is your amplifier.   
Good media handling:

    one official spokesperson
    press briefings
    regular updates
    no guessing
    
Bad handling:
    
    multiple conflicting statements
    ignoring journalists
    deleting criticism (looks like cover-up)
