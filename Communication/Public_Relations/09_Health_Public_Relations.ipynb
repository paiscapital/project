# Health Public Relations

Create trust between healthcare organizations and society.  
Why does Health PR exist? Imagine this:

    A hospital discovers a new cancer treatment.
    Scientist says:
        "This treatment increases apoptotic pathways by activating mitochondrial cytochrome c release."

    Normal people:
        "...What?"

    Nobody understands.
    Nobody trusts it.
    Nobody uses it.

    Health PR changes it into:
        "This treatment helps the body's natural defense destroy cancer cells more effectively, giving patients a better chance of recovery."

Same science. Different communication. Health PR is the translator.   

## 1: Difference
Medical Marketing = More sales   
Public Health Communication = Improve society's health + healthier population    
Health PR = Build trust 

## 2: Health PR Benefits
#### a) Saving lives
People only follow health advice if they believe it. COVID showed this:

    Countries with higher trust often had higher compliance.
    Communication itself becomes a medical intervention.

#### b) Changing behaviour
Most diseases today are lifestyle diseases:

    Smoking
    Obesity
    Diabetes
    Hypertension
    Poor diet
    Exercise

Medicine alone cannot solve these. People must change behaviour. Health PR helps people change habits.  

#### c) Protecting institutions
Hospitals survive because people trust them. 

    One scandal, 
    One fake news, 
    One misinformation campaign

Can destroy decades of reputation. Health PR protects that reputation

## 3: Functions of Health PR

1. Public Health Campaigns
2. Crisis Communication
3. Promote healthcare services
4. Support health policy
5. Build trust in research

## 4: HIPAA vs PDPA law
These laws protect people's information. Example, Hospital accidentally posts:

    Patient name
    Disease
    Phone number
    edical history
Huge legal problem, Malaysia PDPA Protects personal data. USA HIPAA Protects health information.  Health PR must know both because communication often includes patient stories, photos, and records

## 5: Theories
These theories explain why people behave the way they do and help PR professionals design effective campaigns.

#### a) Health Belief Model (HBM)
People take health action only if they believe:

1. They are at risk (susceptibility).
2. The disease is serious (severity).
3. The action will help (benefit).
4. The obstacles are manageable (barriers).
5. They receive a reminder (cue to action).
6. They believe they can do it (self-efficacy).

#### b) Social Marketing Theory (4Ps)
Instead of selling products, you're "selling" healthy behavior:

    Product: The healthy behavior (e.g., washing hands).
    Price: Reduce cost or inconvenience.
    Place: Make it easy to access (schools, clinics).
    Promotion: Communicate through media, community outreach, and digital channels.

#### C) Two-Way Symmetrical Model
Don't just talk to the public, Listen.  
  
The organization learns from the public, and the public learns from the organization.  

Example: A hospital holds a community forum about vaccine concerns, answers questions honestly, and changes its communication based on feedback. This builds lasting trust.  

#### d) Risk Communication Theory
Used during emergencies such as pandemics, disease outbreaks, or contaminated food incidents, The principles are:

    Be fast.
    Be accurate.
    Be transparent about uncertainty.
    Reduce fear without hiding facts.
    Correct misinformation quickly.
Good risk communication prevents panic while encouraging protective action.

#### e) Agenda-Setting and Framing

1. Agenda-Setting: Helps decide what the public thinks is important. Frequent media coverage of dengue makes dengue a priority issue.
 
2. Framing: Shapes how people think about an issue. Obesity can be framed as an individual's responsibility or as a public policy issue involving food environments and education. Different frames influence different solutions.

## 6: Skillset
A top Health PR professional should be able to:

    Translate complex medical information into simple language.
    Handle crises calmly and strategically.
    Understand healthcare regulations and ethics.
    Communicate across cultures and languages.
    Work with doctors, scientists, policymakers, and journalists.
    Show empathy while remaining accurate and evidence-based.

Health Public Relations is the strategic management of communication, trust, reputation, and relationships in healthcare so that people understand medical information, make healthier decisions, and continue to trust health institutions.
